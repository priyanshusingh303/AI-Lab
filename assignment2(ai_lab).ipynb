```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
fig, axes = plt.subplots(5, 2, figsize=(3.5,9))
for digit, ax in enumerate(axes.ravel()):
    index = list(digits.target).index(digit)
    ax.imshow(digits.images[index], cmap='pink')
    ax.set_title(f"{digit}")
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

# Load the dataset
digits = load_digits()

# Display dataset information
print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Labels:", digits.target_names)

# Display the first image
plt.figure(figsize=(4, 4))
plt.imshow(digits.images[0], cmap='pink')
plt.title(f"Label: {digits.target[0]}")
plt.axis('off')
plt.show()
```
